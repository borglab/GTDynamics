```python
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path
data_dir = Path('../../data')
prefix = 'cable_robot_kinodynamics'
```

```python
def load(file_path):
    data = np.loadtxt(file_path, delimiter=',', skiprows=1)
    cable_lengths = data[:, :4]
    cable_velocities = data[:, 4:8]
    cable_accelerations = data[:, 8:12]
    motor_torques = data[:, 12:16]
    positions = data[:, 16:19]
    quaternions_xyzw = data[:, 19:23]
    return cable_lengths, cable_velocities, cable_accelerations, motor_torques, positions, quaternions_xyzw


_, _, _, _, initial_positions, _ = load(data_dir / f'{prefix}_soft_initial_traj.csv')
_, _, _, _, soft_positions, _ = load(data_dir / f'{prefix}_soft_traj.csv')
_, _, _, _, manifold_positions, _ = load(data_dir / f'{prefix}_cm_f_traj.csv')
```

```python
plt.plot(soft_positions[:, 0], soft_positions[:, 1], 'r-')
plt.plot(manifold_positions[:, 0], manifold_positions[:, 1], 'k.')
```
